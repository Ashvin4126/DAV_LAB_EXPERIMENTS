```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm

# -----------------------------------------
# Create Sample Diabetes Dataset
# -----------------------------------------
uci_diabetes = pd.DataFrame({
    "Glucose": [148, 85, 183, 89, 137, 116, 78, 115, 197, 125,
                99, 168, 145, 110, 132, 105, 175, 160, 140, 118],
    "BMI": [33.6, 26.6, 23.3, 28.1, 43.1, 25.6, 31.0, 35.3, 30.5,
            27.5, 29.4, 32.8, 34.2, 24.7, 31.8, 26.3, 36.5, 33.0,
            28.7, 27.2]
})

# Display Dataset
print("Sample Diabetes Dataset:")
print(uci_diabetes.head())

# -----------------------------------------
# Plot Normal Curves
# -----------------------------------------
plt.figure(figsize=(12, 5))

# ---------------- Glucose ----------------
plt.subplot(1, 2, 1)

sns.histplot(
    uci_diabetes["Glucose"],
    bins=8,
    kde=True,
    stat="density",
    color="skyblue"
)

x = np.linspace(
    uci_diabetes["Glucose"].min(),
    uci_diabetes["Glucose"].max(),
    100
)

plt.plot(
    x,
    norm.pdf(
        x,
        uci_diabetes["Glucose"].mean(),
        uci_diabetes["Glucose"].std()
    ),
    color="red",
    linewidth=2
)

plt.title("Normal Distribution - Glucose")
plt.xlabel("Glucose")
plt.ylabel("Density")

# ---------------- BMI ----------------
plt.subplot(1, 2, 2)

sns.histplot(
    uci_diabetes["BMI"],
    bins=8,
    kde=True,
    stat="density",
    color="lightgreen"
)

x = np.linspace(
    uci_diabetes["BMI"].min(),
    uci_diabetes["BMI"].max(),
    100
)

plt.plot(
    x,
    norm.pdf(
        x,
        uci_diabetes["BMI"].mean(),
        uci_diabetes["BMI"].std()
    ),
    color="red",
    linewidth=2
)

plt.title("Normal Distribution - BMI")
plt.xlabel("BMI")
plt.ylabel("Density")

plt.tight_layout()
plt.show()
```
