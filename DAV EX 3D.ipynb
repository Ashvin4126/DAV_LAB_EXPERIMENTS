```python
# Import Libraries
import pandas as pd

# ---------------------------------------
# Create Sample Univariate Statistics
# ---------------------------------------

uci_stats = pd.DataFrame({
    "Feature": [
        "Glucose",
        "BloodPressure",
        "BMI",
        "Age"
    ],
    "Mean": [127.3, 67.0, 30.9, 35.9],
    "Median": [120.5, 68.0, 30.8, 32.5],
    "Standard Deviation": [40.2, 15.4, 6.8, 10.5]
})

pima_stats = pd.DataFrame({
    "Feature": [
        "Glucose",
        "BloodPressure",
        "BMI",
        "Age"
    ],
    "Mean": [133.5, 75.5, 29.4, 36.0],
    "Median": [135.0, 75.0, 30.0, 36.5],
    "Standard Deviation": [28.6, 7.3, 4.2, 8.6]
})

# ---------------------------------------
# Display Statistics
# ---------------------------------------

print("Comparison of Univariate Analysis Results")

print("\nUCI Diabetes Dataset Statistics")
print(uci_stats)

print("\nPima Diabetes Dataset Statistics")
print(pima_stats)

# ---------------------------------------
# Regression Performance
# ---------------------------------------

uci_r2 = 0.82
pima_r2 = 0.79

uci_accuracy = 85.4
pima_accuracy = 81.7

print("\nRegression Model Comparison")
print("------------------------------------")
print(f"UCI Linear Regression R² Score      : {uci_r2}")
print(f"Pima Linear Regression R² Score     : {pima_r2}")
print(f"UCI Logistic Regression Accuracy    : {uci_accuracy}%")
print(f"Pima Logistic Regression Accuracy   : {pima_accuracy}%")

# ---------------------------------------
# Conclusion
# ---------------------------------------

if uci_r2 > pima_r2:
    print("\nLinear Regression: UCI dataset performs better.")
else:
    print("\nLinear Regression: Pima dataset performs better.")

if uci_accuracy > pima_accuracy:
    print("Logistic Regression: UCI dataset performs better.")
else:
    print("Logistic Regression: Pima dataset performs better.")
```
