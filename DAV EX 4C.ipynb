```python
# Import Libraries
import pandas as pd
from scipy.stats import ttest_ind

# -------------------------------
# Create Sample UCI Diabetes Dataset
# -------------------------------
uci_diabetes = pd.DataFrame({
    "Glucose": [148, 85, 183, 89, 137, 116, 78, 115, 197, 125],
    "BloodPressure": [72, 66, 64, 66, 40, 74, 50, 72, 70, 96],
    "BMI": [33.6, 26.6, 23.3, 28.1, 43.1, 25.6, 31.0, 35.3, 30.5, 27.5]
})

# -------------------------------
# Create Sample Pima Diabetes Dataset
# -------------------------------
pima_diabetes = pd.DataFrame({
    "Glucose": [120, 140, 110, 95, 160, 180, 130, 145, 155, 100],
    "BloodPressure": [70, 80, 68, 72, 76, 90, 74, 78, 82, 65],
    "BMI": [25.5, 29.8, 24.3, 26.7, 32.1, 35.5, 30.2, 31.0, 34.8, 23.9]
})

# Columns to compare
numerical_columns = ["Glucose", "BloodPressure", "BMI"]

# -------------------------------
# Perform Independent T-Test
# -------------------------------
t_test_results = {}

for col in numerical_columns:
    t_stat, p_value = ttest_ind(
        uci_diabetes[col],
        pima_diabetes[col],
        equal_var=False
    )

    t_test_results[col] = {
        "T-statistic": round(t_stat, 4),
        "P-value": round(p_value, 4)
    }

# Convert to DataFrame
t_test_df = pd.DataFrame(t_test_results).T

# Display Results
print("Independent T-Test Results")
print("--------------------------------")
print(t_test_df)

# -------------------------------
# Interpretation
# -------------------------------
alpha = 0.05

print("\nInterpretation")
print("--------------------------------")

for col in numerical_columns:
    if t_test_results[col]["P-value"] < alpha:
        print(f"{col}: Significant difference between the two datasets.")
    else:
        print(f"{col}: No significant difference between the two datasets.")
```
