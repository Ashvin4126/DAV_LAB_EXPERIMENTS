```python
# Import Libraries
import pandas as pd
from statsmodels.stats.weightstats import ztest

# -------------------------------
# Create Sample Diabetes Dataset
# -------------------------------
uci_diabetes = pd.DataFrame({
    "Glucose": [
        148, 85, 183, 89, 137,
        116, 78, 115, 197, 125,
        99, 168, 145, 110, 132,
        105, 175, 160, 140, 118
    ]
})

# Display Dataset
print("Sample Diabetes Dataset:")
print(uci_diabetes)

# -------------------------------
# Perform Z-Test
# H0: Mean Glucose = 100
# H1: Mean Glucose ≠ 100
# -------------------------------
z_stat, p_value = ztest(uci_diabetes["Glucose"], value=100)

# Display Results
print("\nZ-Test Results")
print(f"Z-Statistic: {z_stat:.4f}")
print(f"P-Value: {p_value:.4f}")

# Interpretation
alpha = 0.05

if p_value < alpha:
    print("\nReject the null hypothesis.")
    print("The mean Glucose level is significantly different from 100.")
else:
    print("\nFail to reject the null hypothesis.")
    print("There is no significant difference in the mean Glucose level from 100.")
```
