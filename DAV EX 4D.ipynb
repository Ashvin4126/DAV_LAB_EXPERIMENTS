```python
# Import Libraries
import pandas as pd
from scipy.stats import f_oneway

# -------------------------------
# Create Sample UCI Diabetes Dataset
# -------------------------------
uci_diabetes = pd.DataFrame({
    "Glucose": [148, 85, 183, 89, 137, 116, 78, 115, 197, 125],
    "BloodPressure": [72, 66, 64, 66, 40, 74, 50, 72, 70, 96],
    "BMI": [33.6, 26.6, 23.3, 28.1, 43.1, 25.6, 31.0, 35.3, 30.5, 27.5]
})

# -------------------------------
# Create Sample Pima Diabetes Dataset
# -------------------------------
pima_diabetes = pd.DataFrame({
    "Glucose": [120, 140, 110, 95, 160, 180, 130, 145, 155, 100],
    "BloodPressure": [70, 80, 68, 72, 76, 90, 74, 78, 82, 65],
    "BMI": [25.5, 29.8, 24.3, 26.7, 32.1, 35.5, 30.2, 31.0, 34.8, 23.9]
})

# Numerical Columns
numerical_columns = ["Glucose", "BloodPressure", "BMI"]

# -------------------------------
# Perform One-Way ANOVA
# -------------------------------
anova_results = {}

for col in numerical_columns:
    f_stat, p_value = f_oneway(
        uci_diabetes[col],
        pima_diabetes[col]
    )

    anova_results[col] = {
        "F-statistic": round(f_stat, 4),
        "P-value": round(p_value, 4)
    }

# Convert Results to DataFrame
anova_df = pd.DataFrame(anova_results).T

# Display Results
print("One-Way ANOVA Results")
print("--------------------------------")
print(anova_df)

# -------------------------------
# Interpretation
# -------------------------------
alpha = 0.05

print("\nInterpretation")
print("--------------------------------")

for col in numerical_columns:
    if anova_results[col]["P-value"] < alpha:
        print(f"{col}: Significant difference between the datasets.")
    else:
        print(f"{col}: No significant difference between the datasets.")
```
