```python
import pandas as pd

# -----------------------------
# Create sample text (CSV) data
# -----------------------------
text_data = {
    "Name": ["Alice", "Bob", "Charlie", "David", None],
    "Age": [23, 25, None, 30, 28],
    "City": ["Chennai", "Delhi", "Mumbai", None, "Bangalore"]
}

text_df = pd.DataFrame(text_data)

# Save as CSV
text_df.to_csv("sample_text.csv", index=False)

# Read CSV
text_df = pd.read_csv("sample_text.csv")


# -----------------------------
# Create sample Excel data
# -----------------------------
excel_data = {
    "Product": ["Laptop", "Mouse", "Keyboard", None, "Monitor"],
    "Price": [55000, 700, None, 1500, 12000],
    "Stock": [10, 50, 35, 20, None]
}

excel_df = pd.DataFrame(excel_data)

# Save as Excel
excel_df.to_excel("sample_excel.xlsx", sheet_name="Sheet1", index=False)

# Read Excel
excel_df = pd.read_excel("sample_excel.xlsx", sheet_name="Sheet1")


# -----------------------------
# Read data from a website
# -----------------------------
web_df = pd.read_csv(
    "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
)

# -----------------------------
# Display data
# -----------------------------
print("CSV Data:")
print(text_df.head())

print("\nExcel Data:")
print(excel_df.head())

print("\nWeb Data:")
print(web_df.head())

# -----------------------------
# Handle missing values
# -----------------------------
text_df.ffill(inplace=True)      # Forward fill
excel_df.bfill(inplace=True)     # Backward fill
web_df.dropna(inplace=True)      # Remove missing values

# -----------------------------
# Save processed data
# -----------------------------
text_df.to_csv("processed_text.csv", index=False)
excel_df.to_excel("processed_excel.xlsx", index=False)

print("\nProcessed files saved successfully!")
```
